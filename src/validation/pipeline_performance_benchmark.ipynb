```python

import time
from pyspark.sql import functions as F

tabla = spark.table(
    "workspace.silver_flights.flights_clean"
)

resultados = []

for cantidad in [100_000, 1_000_000, 5_000_000]:
    inicio = time.perf_counter()

    resumen = (
        tabla.limit(cantidad)
        .groupBy("delayed_15")
        .count()
        .collect()
    )

    segundos = time.perf_counter() - inicio

    resultados.append((cantidad, round(segundos, 3)))

    print(
        f"Registros: {cantidad:,} | "
        f"Tiempo: {segundos:.3f} segundos"
    )

print("\nBENCHMARK TERMINADO")
```
